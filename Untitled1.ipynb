```python
import json
import pandas as pd


caminho_arquivo = '/content/TelecomX_Data (1).json'


with open(caminho_arquivo, 'r') as f:
    dados_json = json.load(f)


df = pd.json_normalize(dados_json)


print(df.head())



```

```python

print(df.info())
```

```python
missing_values = df.isnull().sum()
print("Valores ausentes:\n", missing_values[missing_values > 0])
```

```python

duplicados = df.duplicated().sum()
print(f"Número de registros duplicados: {duplicados}")
```

```python
for col in df.columns:
    if df[col].dtype == 'object':
        df[col] = df[col].astype(str).str.strip()


df['account.Charges.Monthly'] = pd.to_numeric(df['account.Charges.Monthly'], errors='coerce')
df['account.Charges.Total'] = pd.to_numeric(df['account.Charges.Total'], errors='coerce')


df = df[df['Churn'].isin(['Yes', 'No'])]


df = df.drop_duplicates()


faltantes = df.isnull().sum()
print("Valores ausentes após limpeza:")
print(faltantes[faltantes > 0])


print("\nResumo do DataFrame após limpeza:")
print(df.info())
```

```python

print(df[df['account.Charges.Total'].isnull()][['customer.tenure', 'account.Charges.Monthly']].describe())


df['account.Charges.Total'].fillna(df['account.Charges.Total'].median(), inplace=True)

```

```python
num_cols = ['customer.tenure', 'account.Charges.Monthly', 'account.Charges.Total']
print(df[num_cols].describe(percentiles=[.1, .25, .5, .75, .9]).round(2))


print("\nCorrelação:\n", df[num_cols].corr())
```
